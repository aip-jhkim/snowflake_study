## Snowflake Notebook 기능 📓 

### 노트북 세션 시작

Snowflake Notebook은 포함된 SQL 및 Python 셀에서 코드를 실행하기 위해 **활성** 컴퓨팅 리소스가 필요합니다. 이러한 컴퓨팅 리소스(나중에 설명할 Snowflake 가상 웨어하우스)는 생성 시점에 Snowflake Notebook에 할당됩니다. 그러나 노트북에 진입할 때 즉시 시작되지는 않습니다. 새로운 `session`은 다음 두 가지 방법 중 하나로 트리거할 수 있습니다.

1. 노트북 인터페이스 오른쪽 상단의 **Start** 버튼을 클릭합니다. 완료까지 몇 초가 소요될 수 있으며, 완료되면 버튼이 **Active**로 바뀝니다.
1. 실행 가능한(SQL 또는 Python) 노트북 셀을 실행해 보세요. 시스템은 해당 셀의 코드를 실행하기 전에 새 세션을 생성합니다.

![세션 시작(이미지)](https://edu-cdev-images.s3.us-west-2.amazonaws.com/ob/ob_start_session.png)

💡 **팁:** 이 버튼 오른쪽에 있는 드롭다운 메뉴에는 세션을 끝내는 `End` 옵션이 있습니다. 노트북 작업을 마쳤으면 노트북 세션을 수동으로 종료하는 것이 좋습니다. 노트북은 설정된 유휴 시간 초과 후 자동으로 종료되지만, 작업 완료 후 수동으로 종료하면 컴퓨팅 비용을 절약할 수 있습니다.


### 노트북 셀 기본 사항 📓 

- **셀 추가**
    - 기존 셀의 하단 테두리에 마우스를 올려놓은 후 원하는 셀 유형의 버튼을 클릭합니다. 

- **셀 유형 변경**
    - 기존 셀에서 셀 유형 옆의 아래쪽 화살표를 선택합니다.

- **셀 삭제**
    - 줄임표를 선택하고 적절한 작업을 선택합니다.

- **셀 복제**
    - 줄임표를 선택하고 적절한 작업을 선택합니다.    

- **셀 이동**
    - 줄임표(세로로 쌓인 세개의 점)를 선택하고 적절한 작업(Move cell up - 셀 위로 이동, Move cell down - 셀 아래로 이동)을 선택합니다.
    - 이동할 셀 위에 마우스를 올린 후 드래그 앤 드롭 아이콘을 선택하고 셀을 새 위치로 이동합니다.

## Notebook 셀 간 Python 변수 치환 🥋

Snowflake Notebook에서는 Jinja 구문 `{{..}}`을 사용하여 이전에 정의된 Python 변수를 이후에 나오는 Python 셀과 SQL 셀에서 참조할 수 있습니다.

이를 시연해 보겠습니다. 먼저 Python 변수를 정의하고, 해당 값을 화면에 출력해 보겠습니다. 그렇게 하려면 다음 Python 셀을 실행합니다.


```
greeting = '이 코스에 오신 것을 환영합니다!'
print(greeting)
```

이렇게 저장된 변수는 이후의 Python 셀에서 다시 사용할 수 있습니다. 다음을 실행합니다.


```
print("이것은 이전 셀에서 정의된 \"greeting\"입니다:")
print(greeting)
```

Jinja 문법을 활용하면 **SQL** 셀에서도 이 변수 치환을 사용할 수 있습니다. 다음을 시도해 보세요.


```
%%sql -r dataframe_2
SELECT '{{greeting}}' AS "이것은 이전 셀에서 정의된 인사말입니다:";
```

## 지식 테스트 :mag_right:

아래의 대화형 퀴즈 문제를 통해 이해도를 확인해 보세요. 각 `RUN_THIS_QUIZ_QUESTION_` 셀에는 Snowflake 기능과 관련된 객관식 문제를 제시하는 Streamlit 위젯이 포함되어 있습니다.  

**지침:**  
1. 노트북 셀 위에 커서를 올려 추가 컨트롤을 표시하세요.
1. 각 퀴즈 셀 오른쪽의 ▶️ **Play 버튼**을 클릭하여 실행하세요.  
1. 제공된 옵션에서 답을 선택하세요.  
1. 다음으로 넘어가기 전에 피드백을 검토하세요.  

💡 **참고:** 궁금하시면 셀을 확장하여 코드를 볼 수 있지만, 필수는 아닙니다. 이 퀴즈들은 필수 사항이 아닙니다. 배운 내용을 복습하며 연습할 기회를 제공하기 위한 것입니다.  


```
import streamlit as st
try:
    if session is None:
        print('session is None')
except NameError:
    from snowflake.snowpark.context import get_active_session
    session = get_active_session()

question = "아래 어떤 문장이 Snowflake 인터페이스에 대해 사실인가요?"
options = ["아래 선택을 고르세요...",
           "A) 이름은 Main과 Notetake입니다", 
           "B) Classic Console과 Snowdrift라고 불립니다", 
           "C) 각각 오브젝트 브라우저를 가지고 있습니다", 
           "D) 메인 인터페이스만 오브젝트 선택기가 있습니다"]

st.divider()
user_answer = st.radio(question, options, index=0)
if user_answer:
    if user_answer == "아래 선택을 고르세요...":
        ''
    else:
        answer = '5a435ed312078d60147fb05fcfde2364'
        # 옵션 문자 (A, B, C, 또는 D)를 추출하세요
        selected_option = user_answer.split(')')[0] + ')'
        response = session.sql(f"call common_db.resources.quiz_temp('{answer}', '{user_answer}', 'False')").collect()
        if response:
            value = response[0]['QUIZ_TEMP']
        st.write(f"{selected_option} {value}")
```
