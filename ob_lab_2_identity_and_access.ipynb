## 데이터베이스 및 스키마 📓

데이터베이스는 데이터 세트(테이블 및 기타 오브젝트)를 그룹화합니다. 데이터베이스 내에서 두 번째 수준의 조직 단위를 **스키마**라고 합니다. 데이터베이스를 생성할 때마다 Snowflake는 **자동으로** 두 개의 스키마를 생성합니다.

1. `INFORMATION_SCHEMA` 스키마는 뷰들의 집합을 포함하고 있습니다. `INFORMATION_SCHEMA` 스키마는 삭제(Drop), 이름 변경 또는 이동할 수 없습니다.

1. `PUBLIC` 스키마는 비어 있는 상태로 생성되며, 시간이 지남에 따라 테이블, 뷰 및 기타 오브젝트로 채울 수 있습니다. `PUBLIC` 스키마는 언제든지 Drop, 이름 변경 또는 이동할 수 있습니다.  


### 새로 만든 데이터베이스에서 스키마에 대한 정보를 표시하는 코드를 실행합니다.

`SHOW` 명령은 Snowflake 내 데이터베이스에 포함된 스키마의 일반 정보를 확인하는 데도 사용할 수 있습니다. 다음을 수행하세요.


```
SHOW SCHEMAS IN DATABASE {{your_db}};
```

- Snowflake 사용 시 오브젝트가 사라진 것처럼 보일 수 있습니다. 생성된 항목임에도 **'Does Not Exist'** 오류가 발생할 수 있습니다. 이 경우 먼저 자신의 역할을 확인하세요.

Snowflake에서는 역할이 오브젝트를 소유하지 않아도 해당 오브젝트를 조회하거나 사용할 수 있습니다. 오브젝트에 대한 특정 작업 권한도 `GRANT` 명령을 사용하여 역할에 위임할 수 있습니다. Snowsight 인터페이스는 오브젝트에 대한 작업을 수행할 수 있는 옵션을 제공합니다. 

다음 Snowsight 화면을 살펴보세요. 

![권한 부여(이미지)](https://edu-cdev-images.s3.us-west-2.amazonaws.com/ob/ob_grant_privileges_1_v2.png)

다음은 잠시 후 이 단계들을 직접 수행하는 데 도움이 될 만한 설명입니다.

- 오브젝트 브라우저에서 스키마를 선택하면, 오른쪽에 Schema details 패널이 표시됩니다.

- `+ Privilege` 버튼을 확인하세요. 이 버튼을 클릭하면 **Grant new privileges on** 대화 상자가 열립니다.

- **Role** 드롭다운은 해당 스키마에 대한 권한을 제공(부여)할 역할의 이름을 선택하는 곳입니다. 이 항목을 클릭하면 **Search roles** 옵션이 나타나며, 이를 사용하여 원하는 역할을 찾을 수 있습니다.

- **Privileges** 드롭다운에는 다양한 오브젝트에 걸친 Snowflake 권한 목록이 포함되어 있습니다.

- **USAGE** 권한은 스키마를 조회하고 사용하기 위한 기본적인 액세스 권한을 제공합니다.

## 지식 테스트 :mag_right:

아래의 대화형 퀴즈 문제를 통해 이해도를 확인해 보세요. 각 `RUN_THIS_QUIZ_QUESTION_` 셀에는 Snowflake 기능과 관련된 객관식 문제를 제시하는 Streamlit 위젯이 포함되어 있습니다.  

**지침:**  
1. 노트북 셀 위에 커서를 올려 추가 컨트롤을 표시하세요.
1. 각 퀴즈 셀 오른쪽의 ▶️ **Play 버튼**을 클릭하여 실행하세요.  
1. 제공된 옵션에서 답을 선택하세요.  
1. 다음으로 넘어가기 전에 피드백을 검토하세요. 

💡 **참고:** 궁금하시면 셀을 확장하여 코드를 볼 수 있지만, 필수는 아닙니다. 이 퀴즈들은 필수 사항이 아닙니다. 배운 내용을 복습하며 연습할 기회를 제공하기 위한 것입니다.  


```
question = "이 실습에서 수행한 작업을 바탕으로 다음 중 데이터베이스 소유권에 대해 참인 것은 무엇인가요?"
options = ["아래 선택을 고르세요...",
           "A) SYSADMIN이 대부분의 데이터베이스를 소유하고 있습니다", 
           "B) SECURITYADMIN이 대부분의 데이터베이스를 소유하고 있습니다", 
           "C) 데이터베이스의 소유권은 처음에 이를 생성한 역할에 있습니다", 
           "D) 데이터베이스를 생성한 후에는 다른 역할로 소유권을 이전할 수 없습니다"]

st.divider()
user_answer = st.radio(question, options, index=0)
if user_answer:
    if user_answer == "아래 선택을 고르세요...":
        ''
    else:
        answer = 'f3117b2957b108688e4a6e01e9664b24'
        # 옵션 문자 (A, B, C, 또는 D)를 추출하세요
        selected_option = user_answer.split(')')[0] + ')'
        response = session.sql(f"call common_db.resources.quiz_temp('{answer}', '{user_answer}', 'False')").collect()
        if response:
            value = response[0]['QUIZ_TEMP']
        st.write(f"{selected_option} {value}")
```
