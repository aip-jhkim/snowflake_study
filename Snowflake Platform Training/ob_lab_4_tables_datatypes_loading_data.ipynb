## Select Star와 LIMIT 📓

Select Star 구문은 '`SELECT *`'로 시작합니다(기술적으로는 별표, 즉 asterisk입니다). 'Select Asterisk'라고 말하면 너무 길죠. Select Star는 각 컬럼을 하나씩 나열하지 않고 테이블의 모든 컬럼을 조회하기 위한 방법입니다.


### 새 테이블에서 Select Star를 실행하는 방법 🥋

다음 SQL 셀의 샘플 코드를 수정하여 새 `ROOT_DEPTH` 테이블의 모든 레코드를 선택하는 쿼리를 작성해 보세요.

- 다음 구문에서 해시(`#`) 문자를 적절한 구문으로 바꾸세요.


```
-- 이 샘플 코드를 수정하세요
SELECT # FROM #;
```

### Select Star의 다양한 형태 📓

Select Star를 사용할 때, Snowflake는 타이핑을 줄이면서도 출력 결과를 맞춤화할 수 있는 여러 가지 편리한 SQL 기능을 제공합니다. 이는 컬럼이 많은 테이블, 테이블 함수, 뷰를 다룰 때 특히 유용합니다. 쿼리를 작성하면서 수십 개, 수백 개의 컬럼 이름을 일일이 입력하는 일은 전혀 즐겁지 않죠. 당신의 시간은 그보다 훨씬 더 소중하니까요.

`SELECT *`를 지정할 때, `EXCLUDE`, `REPLACE`, `RENAME` 및 `ILIKE` 옵션을 포함할 수 있습니다.

- `EXCLUDE`: 제외 대상으로 지정된 컬럼을 제외한 나머지 모든 컬럼을 반환합니다.
- `REPLACE`: 모든 컬럼을 반환하되, 지정된 컬럼의 값을 계산된 표현식의 값으로 대체합니다.
- `RENAME`: 모든 컬럼을 반환하며, 결과에 사용할 컬럼 별칭을 지정합니다.
- `ILIKE`: 지정된 패턴과 일치하는 모든 컬럼을 반환합니다.


### 아래 예시들을 새로 만든 테이블에서 실행해 보며, 이러한 축약형 변형을 사용해 컬럼 출력 결과를 조정해 보세요.


```
-- 제외 목록에 있는 컬럼을 제외한 테이블의 모든 컬럼을 반환하려면 EXCLUDE를 사용하세요

SELECT * EXCLUDE (range_min, range_max)
FROM ROOT_DEPTH;
```

```
-- 테이블의 모든 컬럼을 반환하고, root_depth_code 컬럼의 값에 '-RDC' 문자열을 추가하여 REPLACE를 사용하세요

SELECT * REPLACE (root_depth_code||'-RDC' AS root_depth_code)
FROM ROOT_DEPTH;
```

```
-- 테이블의 모든 컬럼을 반환하고, RENAME을 사용하여 범위 컬럼의 제목/헤더를 변경합니다.

SELECT * RENAME (range_min AS root_depth_range_min, 
                 range_max AS root_depth_range_max)
FROM ROOT_DEPTH;
```

```
-- 테이블에서 문자열 `ROOT`를 포함하는 모든 컬럼을 반환하려면 ILIKE를 사용하세요

SELECT * ILIKE '%ROOT%'
FROM ROOT_DEPTH;
```

### Select Star와 LIMIT에 대해 알아보기 📓  

모든 컬럼은 필요하지만 모든 로우가 필요하지 않다면, `LIMIT`를 사용한 Select Star를 실행할 수 있습니다. LIMIT는 소수의 로우만 조회하도록 보장해 줍니다. 이렇게 하면 수백만 개의 로우가 있더라도, 몇 개의 로우만 확인하고 싶을 때 불필요하게 모든 데이터를 불러오며 컴퓨팅 자원을 낭비하지 않게 됩니다.   

물론 현재 우리의 테이블에는 아직 한 로우만 있지만, 이후에는 더 많은 로우가 추가될 것이고, 앞으로 생성하게 될 다른 테이블들에는 훨씬 더 많은 데이터가 로드될 것입니다.


```
SELECT *
FROM ROOT_DEPTH
LIMIT 1;
```
