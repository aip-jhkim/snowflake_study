## 실행될 내용과 실행된 내용 파악하기 📓 

노트북 셀에서는 Python이든 SQL이든 관계없이, 하나의 셀 안에 작성된 **모든** 구문은 셀을 실행할 때 하나의 블록으로 함께 실행됩니다. 이 때문에 여러 개의 구문이 있을 경우, 오류가 발생하지 않는 한 화면에 표시되는 결과는 마지막 구문의 실행 결과가 됩니다.

셀 실행 상태는 셀에 표시되는 색상으로 확인할 수 있습니다. 이 상태 색상은 두 곳에 표시됩니다. 하나는 셀 왼쪽 가장자리, 다른 하나는 오른쪽 셀 탐색 맵입니다.

**셀 상태 색상**:

- **파란색 점** - 셀이 수정되었으나 아직 실행되지 않았음을 나타냅니다.

- **빨간색** - 오류가 발생했습니다.

- **녹색** - 실행이 성공했습니다.

- **움직이는 녹색** - 셀이 현재 실행 중입니다.

- **깜박이는 회색** - 셀이 실행 대기 중입니다. 이 상태는 여러 셀이 동시에 실행되도록 트리거되었을 때 발생합니다.

**참고**: Markdown 셀은 상태를 표시하지 않습니다.

## 지식 테스트 :mag_right:

아래의 대화형 퀴즈 문제를 통해 이해도를 확인해 보세요. 각 `RUN_THIS_QUIZ_QUESTION_` 셀에는 Snowflake 기능과 관련된 객관식 문제를 제시하는 Streamlit 위젯이 포함되어 있습니다.  

**지침:**  
1. 노트북 셀 위에 커서를 올려 추가 컨트롤을 표시하세요.
1. 각 퀴즈 셀 오른쪽의 ▶️ **Play 버튼**을 클릭하여 실행하세요.  
1. 제공된 옵션에서 답을 선택하세요.  
1. 다음으로 넘어가기 전에 피드백을 검토하세요.

```
st.divider()
question = "다음 중 노트북 셀 상태 표시기에 대한 설명으로 옳은 것은 무엇인가요?"
options = ["아래 선택을 고르세요...",
           "A) 파란색 점 - 셀이 수정되지 않았음을 나타냅니다", 
           "B) 빨강 - 오류가 발생했습니다", 
           "C) 녹색 - 셀이 현재 실행 중입니다",
           "D) 깜빡이는 회색 - 셀이 일시 중지되고 사용자 입력을 기다리고 있음"]

user_answer = st.radio(question, options, index=0)
if user_answer:
    if user_answer == "아래 선택을 고르세요...":
        ''
    else:
        answer = '7b3449b0399cbaaaab9c33d1e0c78b90'
        # 옵션 문자 (A, B, C, 또는 D)를 추출하세요
        selected_option = user_answer.split(')')[0] + ')'
        response = session.sql(f"call common_db.resources.quiz_temp('{answer}', '{user_answer}', 'False')").collect()
        if response:
            value = response[0]['QUIZ_TEMP']
        st.write(f"{selected_option} {value}")
```
